```python
import os
import sys
import time
import torch
import pandas as pd
from torch.utils.data import DataLoader
from datasets import load_dataset
from huggingface_hub import HfApi, login
from sklearn.metrics import accuracy_score

# Add current directory to path
sys.path.insert(0, '/kaggle/working/project_files')

# Import local modules
from data_pretrained import PretrainedDataset
from models_baselines import TwoStream_Spatial_ResNet50

# ==========================================
# CONFIGURATION
# ==========================================
HF_TOKEN = "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"
REPO_ID = "srijan2715/reproductioncheckpoints"
os.environ["HF_HUB_DISABLE_XET"] = "1"

BATCH_SIZE = 16  # Reduced for stability
EPOCHS = 10
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ==========================================
# THE 3 RUNS
# ==========================================
runs = [
    {'seed': 42, 'name': 'twostream_brisc_seed42'},
    {'seed': 43, 'name': 'twostream_brisc_seed43'},
    {'seed': 44, 'name': 'twostream_brisc_seed44'},
]

# ==========================================
# HELPER FUNCTIONS
# ==========================================
def safe_upload(local_path, repo_path):
    """Fail-safe upload to Hugging Face"""
    try:
        api = HfApi()
        api.upload_file(
            path_or_fileobj=local_path,
            path_in_repo=repo_path,
            repo_id=REPO_ID,
            repo_type="model"
        )
        print(f"    ☁️ Uploaded: {repo_path}")
        return True
    except Exception as e:
        print(f"    ⚠️ Upload failed: {e}")
        print(f"    (File saved locally at {local_path})")
        return False

def train_run(run_config):
    print(f"\n{'='*60}")
    print(f"🚀 STARTING: {run_config['name']}")
    print(f"{'='*60}")
    
    # Set seed
    torch.manual_seed(run_config['seed'])
    
    # Load data
    print("📥 Loading BRISC dataset...")
    raw_train = load_dataset('srijan2715/brisc2025-paper-pipeline', split="train")
    raw_val = load_dataset('srijan2715/brisc2025-paper-pipeline', split="val")
    
    train_ds = PretrainedDataset(raw_train, is_training=True)
    val_ds = PretrainedDataset(raw_val, is_training=False)
    
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)

    # Load model
    model = TwoStream_Spatial_ResNet50(num_classes=4).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)
    criterion = torch.nn.CrossEntropyLoss()

    # Training loop
    best_acc = 0.0
    logs = []
    
    for epoch in range(EPOCHS):
        model.train()
        running_loss = 0.0
        correct = 0
        total = 0
        
        for batch in train_loader:
            x, y = batch['image'].to(DEVICE), batch['label'].to(DEVICE)
            optimizer.zero_grad()
            out = model(x)
            loss = criterion(out, y)
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item()
            _, preds = torch.max(out, 1)
            correct += (preds == y).sum().item()
            total += y.size(0)
            
        train_acc = correct / total
        
        # Validation
        model.eval()
        val_correct = 0
        val_total = 0
        with torch.no_grad():
            for batch in val_loader:
                x, y = batch['image'].to(DEVICE), batch['label'].to(DEVICE)
                out = model(x)
                _, preds = torch.max(out, 1)
                val_correct += (preds == y).sum().item()
                val_total += y.size(0)
                
        val_acc = val_correct / val_total
        logs.append({'epoch': epoch+1, 'train_acc': train_acc, 'val_acc': val_acc})
        
        print(f"Epoch {epoch+1}/{EPOCHS} | Train Acc: {train_acc:.4f} | Val Acc: {val_acc:.4f}")
        
        if val_acc > best_acc:
            best_acc = val_acc
            pth_name = f"{run_config['name']}_best.pth"
            torch.save(model.state_dict(), pth_name)
            print(f"    💾 Saved best model: {pth_name}")

    # Save logs
    csv_name = f"{run_config['name']}_logs.csv"
    pd.DataFrame(logs).to_csv(csv_name, index=False)
    print(f"    📝 Saved logs: {csv_name}")

    # Auto-upload
    print(f"\n📤 Uploading to Hugging Face...")
    safe_upload(pth_name, f"baselines/twostream/brisc/{pth_name}")
    safe_upload(csv_name, f"baselines/twostream/brisc/{csv_name}")
    
    print(f"\n✅ COMPLETED: {run_config['name']}")

# ==========================================
# EXECUTE
# ==========================================
if __name__ == "__main__":
    print("🔐 Logging into Hugging Face...")
    login(token=HF_TOKEN)
    print("✅ Login successful!")
    
    start_time = time.time()
    
    for run in runs:
        try:
            train_run(run)
        except Exception as e:
            print(f"❌ CRITICAL ERROR in {run['name']}: {e}")
            print("Moving to next seed...")
    
    end_time = time.time()
    print(f"\n{'='*60}")
    print(f"🏁 ALL 3 TWOSTREAM BRISC RUNS COMPLETE!")
    print(f"⏱️ Total Time: {(end_time - start_time)/3600:.2f} hours")
    print(f"{'='*60}")
```
